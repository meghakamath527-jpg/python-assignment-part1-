```python
# Task 1 — Data Parsing & Profile Cleaning 
from ast import Pass
from unicodedata import numeric


raw_students = [
    {"name": "  ayesha SHARMA  ", "roll": "101", "marks_str": "88, 72, 95, 60, 78"},
    {"name": "ROHIT verma",       "roll": "102", "marks_str": "55, 68, 49, 72, 61"},
    {"name": "  Priya Nair  ",    "roll": "103", "marks_str": "91, 85, 88, 94, 79"},
    {"name": "karan MEHTA",       "roll": "104", "marks_str": "40, 55, 38, 62, 50"},
    {"name": " Sneha pillai ",    "roll": "105", "marks_str": "75, 80, 70, 68, 85"},
]

for student in raw_students: 
    # Clean the name
    clean_name = student["name"].strip().title()
   
    # Convert roll number to integer
    student["roll"] = int(student["roll"])
    
    # Convert marks string to a list of integers
    student["marks"] = [int(m) for m in student["marks_str"].split(", ")]

    if clean_name.replace(" ", "").isalpha():
        print("✓ Valid name")
    else:
        print("✗ Invalid name")

    print(f"================================")
    print(f"Student: {clean_name}")
    print(f"Roll: {student['roll']}")
    print(f"Marks: {student['marks']}")
    print(f"================================\n")

# Requirement for Roll 103
for student in raw_students:
    if student["roll"] == 103:
        print(f"Student: {student['name']}")
        print(f"Roll: {student['roll']}")
        print(f"UPPERCASE Name: {student['name'].upper()}")
        print(f"lowercase Name: {student['name'].lower()}")
```

```python
# Task 2 — Marks Analysis Using Loops & Conditionals

student_name = "Ayesha Sharma"
subjects     = ["Math", "Physics", "CS", "English", "Chemistry"]
marks        = [88, 72, 95, 60, 78]

print(f"================================\n")

# Q1-print each subject alongside its marks and a grade label

for i in range(len(subjects)):
    subject = subjects[i]
    mark = marks[i]
    
    if mark >= 90:
        grade = "A+"
    elif mark >= 80:
        grade = "A"
    elif mark >= 70:
        grade = "B"
    elif mark >= 60:
        grade = "C"
    else:
        grade = "F"
    
    print(f"{subject}: {mark} - Grade: {grade}")


print(f"================================\n")

#Q2 Calculate and print:

# - Total Marks
total_marks = sum(marks)
print(f"Total Marks: {total_marks}")
# - Average Marks
average_marks = total_marks / len(marks)
print(f"Average Marks: {average_marks:.2f}")
# - Highest Mark
highest_mark = max(marks)
print(f"Highest Mark: {highest_mark}")
# - Lowest Mark
lowest_mark = min(marks)
print(f"Lowest Mark: {lowest_mark}")
print(f"================================\n")


#Q3 simulate a marks-entry system that allows adding new subjects:

                             
new_subjects_count = 0
while True:
    new_subject = input("Enter a new subject name (or type 'done' to finish): ")
    
    if new_subject.lower() == "done":
        break
    
    try:
        new_mark = int(input(f"Enter marks for {new_subject} (0-100): "))
        
        if 0 <= new_mark <= 100:
            subjects.append(new_subject)
            marks.append(new_mark)
            new_subjects_count += 1
        else:
            print("⚠️ Invalid marks! Please enter a number between 0 and 100.")
    except ValueError:
        print("⚠️ Invalid input! Please enter a numeric value for marks.")

print(f"================================")
print(f"Number of new subjects added: {new_subjects_count}")

# Check if we only added ONE new thing
if new_subjects_count == 1:
    # Print the last item added to the list
    print(f"New Subject Mark: {marks[-1]}") 
else:
    # Calculate average of everything
    avg = sum(marks) / len(marks)
    print(f"Updated Total Average: {avg:.2f}")

print(f"================================")

```

```python
#Task 3 — Class Performance Summary

class_data = [
    ("Ayesha Sharma",  [88, 72, 95, 60, 78]),
    ("Rohit Verma",    [55, 68, 49, 72, 61]),
    ("Priya Nair",     [91, 85, 88, 94, 79]),
    ("Karan Mehta",    [40, 55, 38, 62, 50]),
    ("Sneha Pillai",   [75, 80, 70, 68, 85]),
]
#Q1. Loop through class_data and for each student:


for student in class_data:
    name = student[0]
    marks = student[1]
    
    average = sum(marks) / len(marks)
    status = "Pass" if average >= 60 else "Fail"
    print(f"Student: {name}")
    print(f"Average Marks: {average:.2f}")
    print(f"Result: {status}")
    print(f"================================\n")

#Q2. Print a formatted class report:


print(f"{'Name':<18} | {'Average':<7} | {'Status'}")
print("-" * 40)
for student in class_data:
    name = student[0]
    marks = student[1]
    
    average = sum(marks) / len(marks)
    status = "Pass" if average >= 60 else "Fail"
    print(f"{name:<18} | {average:<7.2f} | {status}")

print(f"================================\n")

#Q3. Number of students who passed and who failed
#Class topper (name + average)
#Class average (average of all five students' averages)

pass_count = 0
fail_count = 0
total_average = 0
for student in class_data:
    marks = student[1]
    average = sum(marks) / len(marks)
    total_average += average
    
    if average >= 60:
        pass_count += 1
    else:
        fail_count += 1
class_average = total_average / len(class_data)
print(f"\nTotal Passed: {pass_count}")
print(f"Total Failed: {fail_count}")
print(f"Class Average: {class_average:.2f}")

print(f"================================\n")
```

```python
#Task 4 — String Manipulation Utility

essay = "  python is a versatile language. it supports object oriented, functional, and procedural programming. python is widely used in data science and machine learning.  "

clean_essay = essay.strip().lower()
print(f"Title Case: {clean_essay.title()}") #Title Case convertion
python_count = clean_essay.count("python")
print(f"Count of 'python': {python_count}")  #Count how many times "python" appears in the essay
replaced_essay = clean_essay.replace("python", "Python 🐍"
                                     )
print(f"Replaced Essay: {replaced_essay}") #Replace all occurrences of "python" with "Python 🐍"
sentences = clean_essay.split(". ")     #Split the essay into sentences based on the period followed by a space
print(f"Sentences: {sentences}")    
for i, sentence in enumerate(sentences, start=1):
    if not sentence.endswith("."):
        sentence += "."
    print(f"{i}. {sentence}")               #Print each sentence on a new line, prefixed with its sentence number (e.g., "1. ...", "2. ...", etc.)

print(f"================================\n")
```
